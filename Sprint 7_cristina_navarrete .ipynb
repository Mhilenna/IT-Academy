1. Creamos una función para obtener datos mediante input: `obtenir_dades()`

Pedimos nombre, teléfono, antiguedad y le pedimos del 0 al 3 el nivel de cataà i excel y finalmente le damos a escoger mediante un print primero las opciones del nivel académico.

```python
def obtenir_dades():
    nom = input("Introdueix el teu nom i cognoms: ")
    telefon = input("Introdueix el telèfon: ")
    antiguitat = input("Quants anys d'experiència tens a l'empresa? ")
    catala = input("Introdueix el nivell de català (0-3): ")
    excel = input("Introdueix el nivell d'Excel (0-3): ")
    print("\n--- NIVELL ACADÈMIC ---")
    print("0 - ESO")
    print("1 - FP")
    print("2 - Universitat")
    print("3 - Màster o superior")
    nivell_academic = input("Selecciona el nivell acadèmic: ")

    resultat = [nom, telefon, antiguitat, catala, excel, nivell_academic]

    return resultat
```

2. Creamos una función para validar los datos y convertirlos: `validar_Dades(resultat)`

La función primero convierte los datos que llegan como texto: el nombre y el teléfono se quedan como texto, pero sin espacios sobrantes; los años de experiencia pasan a número decimal (cambiando antes la coma por un punto, para que “2,5” valga 2.5); y los niveles de català, Excel y estudios pasan a números enteros. Después comprueba que todo tiene sentido, y solo acepta al candidato si se cumplen todas las condiciones a la vez: que el nombre no esté vacío, que el teléfono tenga solo dígitos, que la experiencia sea 0 o más, y que los tres niveles estén entre 0 y 3.

```python
def validar_Dades(resultat):
    try:
        nom = str(resultat[0]).strip()
        telefon = str(resultat[1]).strip()
        antiguitat = float(str(resultat[2]).replace(",", "."))
        catala = int(resultat[3])
        excel = int(resultat[4])
        nivell_academic = int(resultat[5])

        if (nom != "" and telefon.isdigit()
                and antiguitat >= 0
                and 0 <= catala <= 3
                and 0 <= excel <= 3
                and 0 <= nivell_academic <= 3):

            return [nom, telefon, antiguitat,
                    catala, excel, nivell_academic]

        else:
            print(f"Dades incorrectes per al candidat: {nom}")
            return None

    except (ValueError, TypeError):
        print("Dades incorrectes: hi ha un valor amb format no vàlid.")
        return None
```

6. Ahora creamos la función para mostrar el informe global ordenado por prioritat: `mostrar_informe(llista_candidats)`

La función mostrar_informe recibe la lista de candidatos, donde cada uno es un diccionario con su nombre, teléfono, puntuación y prioridad, y la ordena con sorted. Para decidir el orden, el lambda lee de cada candidato su prioridad, que se convierte en un número con el diccionario mapeo_prioritat (Alta = 1, adequat = 2, No prioritari = 3), y su puntuación con el signo cambiado. Python ordena primero por ese número de prioridad y, si dos candidatos empatan, por la puntuación, de modo que quien tiene más puntos sale antes. Después imprime el título y recorre la lista ya ordenada, mostrando el nombre, el teléfono, la puntuación y la prioridad de cada candidato, y devuelve esa lista ordenada.

```python
def mostrar_informe(llista_candidats):
    ordre_prioritat = {
        "Prioritat Alta": 1,
        "Candidat adequat": 2,
        "No prioritari": 3}

    informe_ordenat = sorted(
        llista_candidats,
        key=lambda candidat: (ordre_prioritat[candidat["prioritat"]], -candidat["puntuacio"]))

    print("\n=== INFORME FINAL DE CANDIDATS ===")
    print(f"{'Nom':<20}{'Telèfon':<14}{'Puntuació':<11}Prioritat")
    print("-" * 60)

    for candidat in informe_ordenat:
        print(f"{candidat['nom']:<20}{candidat['telefon']:<14}{candidat['puntuacio']:<11}{candidat['prioritat']}")

    return informe_ordenat
```

1. Función para obtener segment, factor i missatge: `obtenir_segment_i_factor(codigo)`

Esta función recibe un código del 0 al 4, busca en el diccionario comportamientos qué comportamiento corresponde a ese código y devuelve un nuevo diccionario con el código, la descripción del comportamiento (segment), su factor numérico (factor) y el mensaje publicitario asociado (missatge).



```python

def obtenir_segment_i_factor(codigo):
    comportamientos = {
        0: {
            "segment": "Navega pero no añade productos",
            "factor": 0.10,
            "missatge": "Descubre nuestros productos"
        },
        1: {
            "segment": "Añade productos pero no llega al checkout",
            "factor": 0.25,
            "missatge": "Completa tu compra"
        },
        2: {
            "segment": "Llega al checkout pero no paga",
            "factor": 0.50,
            "missatge": "Finaliza el pago"
        },
        3: {
            "segment": "Cesta abandonada hace menos de 24 h",
            "factor": 0.75,
            "missatge": "Aún estás a tiempo de completar la compra"
        },
        4: {
            "segment": "Cesta abandonada hace más de 24 h",
            "factor": 0.90,
            "missatge": "Recupera tu cesta de compra"
        }
    }

    return {
        "codigo": codigo,
        "segment": comportamientos[codigo]["segment"],
        "factor": comportamientos[codigo]["factor"],
        "missatge": comportamientos[codigo]["missatge"]
    }
```

2. Funció per obtenir_factor_edat: `obtenir_factor_edat(edat)`

Esta función recibe una edad (edat) y recorre el diccionario rangs para comprobar en qué intervalo está. Cuando encuentra el intervalo, devuelve su factor; si no encuentra ninguno, devuelve 'Edat no vàlida'.

```python

def obtenir_factor_edat(edat):
    rangs = {
        range(18, 25): 1.20,
        range(26, 40): 1.00,
        range(41, 60): 0.85,
        range(61, 121): 0.70}

    for rang, factor in rangs.items():
        if edat in rang:
            return factor

    return 'Edat no vàlida'
```

3. Función para calcular el riesgo: `calcular_risc(factor_codi, factor_edat)`

Esta función recibe dos factores, factor_codi y factor_edat, los multiplica y guarda el resultado en risc_final. Después devuelve ese resultado con return.







```python
def calcular_risc(factor_codi, factor_edat):
  risc_final = factor_codi * factor_edat
  return risc_final
```

4.Funció per processar clients: `processar_clients(clients)`

Esta función recibe una lista de clientes y crea un diccionario vacío llamado resultats. Después recorre los clientes uno a uno: obtiene el segmento, el factor y el mensaje de su comportamiento, busca el factor correspondiente a su edad y calcula el riesgo. Guarda los datos y resultados de cada cliente usando su identificador como clave. Al terminar, devuelve el diccionario con todos los clientes procesados.




```python
def processar_clients(clients):
    resultats = {}

    for client in clients:
        dades_comportament = obtenir_segment_i_factor(
            client["codi_comportament"])

        factor_edat = obtenir_factor_edat(client["edat"])

        risc = calcular_risc(
        dades_comportament["factor"],
        factor_edat)

        resultats[client["id_client"]] = {
            "id_cliente": client["id_client"],
            "nombre": client["nom"],
            "edad": client["edat"],
            "grupo_edad": client["grup_edat"],
            "codigo_comportamiento": client["codi_comportament"],
            "segment": dades_comportament["segment"],
            "missatge": dades_comportament["missatge"],
            "factor_codi": dades_comportament["factor"],
            "factor_edat": factor_edat,
            "risc_final": risc}

    return resultats

```

1. Funció per netejar dades: `netejar_dades(lista_inicial)`

Esta función recibe lista_inicial y limpia cada texto: convierte las letras a minúsculas con .lower(), cambia | por comas, elimina los espacios que hay justo antes y después de las comas y sustituye los demás espacios por guiones bajos _.

Guarda los textos modificados en una nueva lista llamada lista_neta y la devuelve con return

```python
def netejar_dades(lista_inicial):
  lista_neta = [texto.lower()
        .replace("|", ",")
        .replace(" ,", ",")
        .replace(", ", ",")
        .replace(" ", "_")
    for texto in lista_inicial]

  return(lista_neta)
```

2. Funció per convertir a diccionari: `conversio_dades(pacients)`

Esta función recorre la lista de pacientes y separa cada texto por las comas con split(","). Extrae el nombre, le quita los espacios de los extremos y lo convierte a minúsculas; convierte la edad en un número entero y guarda los datos restantes como una lista de síntomas. Después guarda cada paciente en un diccionario usando su nombre como clave y devuelve el diccionario completo.

```python
def conversio_dades(pacients):
    diccionari_pacients = {}

    for pacient in pacients:
        dades = pacient.split(",")

        nom = dades[0].strip().lower()
        edat = int(dades[1])
        simptomes = dades[2:]


        diccionari_pacients[nom] = {
            "edat": edat,
            "simptomes": simptomes
        }

    return diccionari_pacients
```

3. Funció per assignar risc als símptomes i afegir el total: `afegir_risc(diccionari_pacient, nivells_risc)`

Esta función recorre los pacientes y busca el valor de riesgo de cada síntoma en nivells_risc. Sustituye la lista de síntomas por un diccionario con cada síntoma y su riesgo, suma esos valores y guarda la suma en "risc_total". Finalmente, devuelve el diccionario de pacientes actualizado.

```python

def afegir_risc(diccionari_pacient, nivells_risc):


    for nom, pacient in list(diccionari_pacient.items()):
        simptomes_amb_risc = {}
        risc_total = 0

        for simptoma in pacient["simptomes"]:
            valor = nivells_risc[simptoma]
            simptomes_amb_risc[simptoma] = valor
            risc_total += valor

        pacient["simptomes"] = simptomes_amb_risc
        pacient["risc_total"] = risc_total

    return diccionari_pacient
```

```python
def ordenar_per_risc (diccionari_pacient):
   diccionari_pacient_ordenat = dict(sorted(diccionari_pacient.items(), key=lambda item: item[1]['risc_total'],reverse = True))
   return diccionari_pacient_ordenat
```

EXERCICI 1: Generador de contrasenyes segures


Ets analista de dades dins el departament de Ciberseguretat d'una empresa tecnològica. S'ha detectat que la majoria de treballadors continuen utilitzant contrasenyes molt dèbils com “Asdf1234”, dates d'aniversari o combinacions fàcils d'endevinar. Per aquest motiu, el responsable del departament t'ha demanat desenvolupar un generador de contrasenyes segures, totalment parametritzable i modular.


Algoritmo paso a paso:

1. seleccionar_caracter(tipo): seleccionar al azar un carácter del tipo indicado.

2. obtenir_generadors(majuscula, minuscula, numeros, signes): crear la lista de tipos de caracteres activados.

3. validar_opcions(longitud, generadors): comprobar que haya algún tipo activado y que la longitud permita incluir uno de cada tipo.

4. generar_caracters(longitud, generadors): generar los caracteres hasta alcanzar la longitud indicada, incluyendo al menos uno de cada tipo activado.

5. convertir_en_contrasenya(caracters): mezclar los caracteres y unirlos en un texto.

6. crear_contrasenya(...): coordinar las funciones anteriores y devolver la contraseña final.



1. Importar las herramientas

random permite elegir y mezclar elementos al azar. string contiene las letras, los números y los signos.

```python
import random
import string
```

2. Seleccionar un carácter `seleccionar_caracter(tipo)`

Esta función recibe un tipo, busca los caracteres correspondientes en el diccionario y devuelve uno al azar.

```python
def seleccionar_caracter(tipo):
    opciones = {
        "majuscula": string.ascii_uppercase,
        "minuscula": string.ascii_lowercase,
        "numeros": string.digits,
        "signes": string.punctuation}

    return random.choice(opciones[tipo])
```

```python
seleccionar_caracter("majuscula")
```

3. Obtener los tipos activados: `obtenir_generadors(majuscula, minuscula, numeros, signes)`

Esta función crea una lista con los nombres de los tipos de caracteres que has activado. Por ejemplo, si solo activas mayúsculas y números, devuelve ["majuscula", "numeros"].

```python
def obtenir_generadors(majuscula, minuscula, numeros, signes):
    generadors = []

    if majuscula:
        generadors.append("majuscula")
    if minuscula:
        generadors.append("minuscula")
    if numeros:
        generadors.append("numeros")
    if signes:
        generadors.append("signes")

    return generadors
```

```python
generadors =obtenir_generadors(True, True, True, False)
obtenir_generadors(True,True,True,False)
```

4. Validar las opciones: `validar_opcions(longitud, generadors)`

Esta función comprueba que hayas activado al menos un tipo de carácter y que la contraseña tenga espacio para incluir uno de cada tipo. Si no se cumple, genera un error con un mensaje explicativo. Si son correctas, termina sin mostrar nada y el programa continúa.

```python
def validar_opcions(longitud, generadors):

    if not generadors:
        raise ValueError("Activa al menos un tipo de carácter")

    if longitud < len(generadors):
        raise ValueError(
            f"La longitud mínima es {len(generadors)}")
```

```python
validar_opcions(8,generadors)
```

5. Generar los caracteres: `generar_caracters(longitud, generadors)`

Esta función añade primero un carácter de cada tipo activado. Después completa la longitud de la contraseña eligiendo tipos y caracteres al azar. Devuelve una lista de caracteres.

```python
def generar_caracters(longitud, generadors):
    caracters = []

    for tipo in generadors:
        caracters.append(seleccionar_caracter(tipo))

    while len(caracters) < longitud:
        tipo = random.choice(generadors)
        caracters.append(seleccionar_caracter(tipo))

    return caracters
```

```python
generar_caracters(8,generadors)
```

6. Formar la contraseña: `convertir_en_contrasenya(caracters)`

Esta función mezcla los caracteres para que no aparezcan siempre en el mismo orden y los une en un único texto.

```python
def convertir_en_contrasenya(caracters):
    random.shuffle(caracters)
    return "".join(caracters)
```

```python
convertir_en_contrasenya(['P', 'y', '3', 'b', 'H', 'i', 'j', '0'])
```

7. Coordinar todo el proceso: `crear_contrasenya(longitud,majuscula=True,minuscula=True,numeros=True,signes=False)`

Esta función llama a las anteriores en orden: obtiene los tipos activados, valida las opciones, genera los caracteres y devuelve la contraseña. Por defecto utiliza mayúsculas, minúsculas y números, pero no signos.

```python
def crear_contrasenya(
    longitud,
    majuscula=True,
    minuscula=True,
    numeros=True,
    signes=False
):
    generadors = obtenir_generadors(
        majuscula, minuscula, numeros, signes
    )

    validar_opcions(longitud, generadors)

    caracters = generar_caracters(longitud, generadors)

    return convertir_en_contrasenya(caracters)
```

2. Funció per separar gols: `separar_gols (partits)`

Esta función separa los goles para poder trabajar con ellos después en dos columnas.

Hacemos copia del DataFrame y separamos en dos columnas: gols_favor y gols_contra. Lo hacemos mediante el split informándole que tiene que separar por "-".

Finalmente eliminamos la columna 'resultat'

```python
def separar_gols (partits):
  partits_gols = partits.copy()
  partits_gols[['gols_favor', 'gols_contra']] = (partits['resultat'].str.split('-', expand=True).astype(int))
  return partits_gols.drop(columns='resultat')
```

3. Funció per agrupar equips: `agrupar_partits(partits_gols)`

Esta función crea un DataFrame con una fila por equipo y partido: separa los datos de locales y visitantes, intercambia los goles a favor y en contra para los visitantes y une ambas tablas. Después indica con un 1 si el equipo ha ganado, empatado o perdido y con un 0 las otras opciones. Calcula los puntos asignando 3 por victoria, 1 por empate y 0 por derrota, y devuelve el DataFrame resultante.

```python
def agrupar_partits(partits_gols):
    local = partits_gols[['local', 'gols_favor', 'gols_contra']].copy()
    local.columns = ['equip', 'gols_favor', 'gols_contra']

    visitant = partits_gols[['visitant', 'gols_contra', 'gols_favor']].copy()
    visitant.columns = ['equip', 'gols_favor', 'gols_contra']

    per_equip = pd.concat([local, visitant], ignore_index=True)

    per_equip['guanyats'] = (
        per_equip['gols_favor'] > per_equip['gols_contra']
    ).astype(int)

    per_equip['empatats'] = (
        per_equip['gols_favor'] == per_equip['gols_contra']
    ).astype(int)

    per_equip['perduts'] = (
        per_equip['gols_favor'] < per_equip['gols_contra']
    ).astype(int)

    per_equip['punts'] = (
    per_equip['guanyats'] * 3
    + per_equip['empatats'] * 1
    + per_equip['perduts'] * 0)

    return per_equip
```

4. Funció per calcular estadístiques: `calcular_estadistiques(per_equip)`

Esta función agrupa los partidos por equipo y suma sus goles a favor, goles en contra, victorias, empates, derrotas y puntos. Después ordena los equipos de mayor a menor puntuación y, en caso de empate, por diferencia de goles. Finalmente, devuelve el DataFrame de la clasificación sin mostrar la columna de diferencia de goles.

```python
def calcular_estadistiques(per_equip):

    taula = per_equip.groupby('equip').agg(
        gols_favor=('gols_favor', 'sum'),
        gols_contra=('gols_contra', 'sum'),
        guanyats=('guanyats', 'sum'),
        empatats=('empatats', 'sum'),
        perduts=('perduts', 'sum'),
        punts=('punts', 'sum')
    )
    taula['diferencia'] = taula['gols_favor'] - taula['gols_contra']
    taula = taula.sort_values(['punts', 'diferencia'], ascending=False)
    return taula.drop(columns='diferencia')

```
